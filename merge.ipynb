```python
cardiovascular.columns = [
    f"{col}-cardiovascular" if col in ["positive", "negative", "very_negative", "neutral", "reasoning"] else col
    for col in cardiovascular.columns
]
```

```python
import html

# Apply html.unescape to the 'menu_title' column
allrecipes['menu_title'] = allrecipes['menu_title'].apply(html.unescape)

# Write the updated DataFrame back to the file
#allrecipes.to_csv('all_recipes.csv', index=False)
```

```python
recipes = pd.merge(allrecipes, cardiovascular_kidney_diabetes, on='menu_title')
recipes.head()
```

```python
len(recipes)
```

```python
recipes.columns
```

```python
len(recipes)
```

```python
recipes['score_cardiovascular'] = (
    recipes['positive-cardiovascular'] - recipes['negative-cardiovascular'] - (2 * recipes['very_negative-cardiovascular'])
) / (
    recipes['positive-cardiovascular'] + recipes['negative-cardiovascular'] + recipes['very_negative-cardiovascular']
)
```

```python
recipes.head()
```

```python
recipes['score_diabetes'] = (
    recipes['positive-diabetes'] - recipes['negative-diabetes'] - (2 * recipes['very_negative-diabetes'])
) / (
    recipes['positive-diabetes'] + recipes['negative-diabetes'] + recipes['very_negative-diabetes']
)
```

```python
recipes.head()
```

```python
recipes['score_kidney'] = (
    recipes['positive-kidney'] - recipes['negative-kidney'] - (2 * recipes['very_negative-kidney'])
) / (
    recipes['positive-kidney'] + recipes['negative-kidney'] + recipes['very_negative-kidney']
)
```
